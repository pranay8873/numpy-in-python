```python
def num(df,column):
    df[column]=pd.to_numeric(df[column],errors="coerce")
    df.loc[df[column]<0,column]=np.nan
    if((df[column].max() - df[column].median())<df[column].mean()):
          df[column]=df[column].fillna(int(df[column].mean()))
    else:
       df[column]=df[column].fillna(int(df[column].median()))
    return df
    
```

```python
tran=(tran.pipe(num,"Budget_Expense"))
```

```python
tran.info()
```

```python
tran=(tran.pipe(num,"Actual_Expense"))
```

```python
tran.info()
```

```python
tran.to_csv("Financial_Transactions_Cleaned.csv", index=False)
```
