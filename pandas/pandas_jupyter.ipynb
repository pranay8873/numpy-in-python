```python
def otype(row):
    if (row["Revenue"]>=30000) & (row["Quantity"]>3):
        return "large bulk order"
    elif row["Revenue"]>30000:
        return "large order"
    else:
        return "small order"
sales["Order type"]=sales.apply(lambda x:otype(x),axis=1)
sales
```
